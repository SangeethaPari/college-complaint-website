# **5.UNIVARIATE ANALYSIS**

```python
category_count = df['Category'].value_counts()

print("Category Distribution:")
print(category_count)

print("\nCategory with Most Courses:")
print(category_count.idxmax())

print("Number of Courses:")
print(category_count.max())

print("\nCategory with Least Courses:")
print(category_count.idxmin())

print("Number of Courses:")
print(category_count.min())
```

```python
platform_count = df['Platform'].value_counts()

print("Platform Distribution:")
print(platform_count)

print("\nDominant Platform:")
print(platform_count.idxmax())

print("Number of Courses:")
print(platform_count.max())

print("\nLeast Represented Platform:")
print(platform_count.idxmin())

print("Number of Courses:")
print(platform_count.min())
```

```python
print("Average Rating:")
print(df['Rating (out of 5)'].mean())

print("\nHighest Rating:")
print(df['Rating (out of 5)'].max())

print("\nLowest Rating:")
print(df['Rating (out of 5)'].min())
```

```python
most_expensive = df.loc[df['Price ($)'].idxmax()]

print("Most Expensive Course")
print("Course Name:", most_expensive['Course_Name'])
print("Price:", most_expensive['Price ($)'])

print("\n")

cheapest = df.loc[df['Price ($)'].idxmin()]

print("Cheapest Course")
print("Course Name:", cheapest['Course_Name'])
print("Price:", cheapest['Price ($)'])
```

```python
print("Average Enrollment:")
print(df['Enrolled_Students'].mean())

highest = df.loc[df['Enrolled_Students'].idxmax()]

print("\nMost Enrolled Course:")
print(highest['Course_Name'])
print("Students:", highest['Enrolled_Students'])

lowest = df.loc[df['Enrolled_Students'].idxmin()]

print("\nLeast Enrolled Course:")
print(lowest['Course_Name'])
print("Students:", lowest['Enrolled_Students'])
```

# **6. BIVARIATE ANALYSIS**

```python
#1. CATEGORY VS ENROLLMENT

category_enrollment = df.groupby('Category')['Enrolled_Students'].mean().sort_values(ascending=False)

print(category_enrollment)

category_enrollment.plot(kind='bar')
plt.title("Average Enrollment by Category")
plt.xlabel("Category")
plt.ylabel("Average Enrollment")
plt.show()
```

```python
#2. PLATFORM VS RATING
platform_rating = df.groupby('Platform')['Rating (out of 5)'].mean().sort_values(ascending=False)

print(platform_rating)

platform_rating.plot(kind='bar')
plt.title("Average Rating by Platform")
plt.xlabel("Platform")
plt.ylabel("Average Rating")
plt.show()
```

```python
#3.CATEGORY VS COMPLETE RATE
category_completion = df.groupby('Category')['Rating (out of 5)'].mean().sort_values(ascending=False)

print(category_completion)

category_completion.plot(kind='bar')
plt.title("Average Completion Rate by Category")
plt.xlabel("Category")
plt.ylabel("Completion Rate")
plt.show()
```

```python
#4.PRICE VS RATING
price_rating = df.groupby('Category')['Rating (out of 5)'].mean().sort_values(ascending=False)

print(price_rating)

price_rating.plot(kind='bar', figsize=(10,5))

plt.title("Average Rating by Category")
plt.xlabel("Category")
plt.ylabel("Average Rating")

plt.show()
```

# **7.CORRELATIONAL ANALYSIS**

```python
numeric_df = df[['Duration (hours)',
                 'Enrolled_Students',
                 'Completion_Rate (%)',
                 'Price ($)',
                 'Rating (out of 5)']]

correlation_matrix = numeric_df.corr()

print(correlation_matrix)
plt.figure(figsize=(8,6))

plt.imshow(correlation_matrix)

plt.colorbar()

plt.xticks(range(len(correlation_matrix.columns)),
           correlation_matrix.columns,
           rotation=90)

plt.yticks(range(len(correlation_matrix.columns)),
           correlation_matrix.columns)

plt.title("Correlation Matrix")

plt.show()
```

### **Interpretation :**

Correlation Analysis was performed to examine the relationships between Duration, Enrolled Students, Completion Rate, Price, and Rating. The correlation coefficients were found to be very close to zero, indicating weak or no linear relationship among the variables. This suggests that factors such as course price, duration, enrollment, and ratings operate independently within the dataset and do not strongly influence one another.

### **Observation:**



*   Duration does not significantly affect enrollment.
*  Duration does not significantly affect completion rate.
*  Price does not significantly affect ratings.
*   Enrollment does not significantly affect ratings.
* No strong positive or negative relationships were observed.
